## 1.Right Angle Triangle Pattern

```python
n = int(input())
for i in range(1,n):
    for j in range(i):
        print('*',end = " ")
    print()
```

## 2. Inverted Right Angle Triangle Pattern

```python
n = int(input())
for i in range(n,0,-1):
    for j in range(i):
        print('*',end = " ")
    print()
```

## 3.Hallow Square Pattern

```python
n = int(input())
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*",end=" ")
        else:
            print(" ", end=" ")
    print()
```

## 4. Full Square Pattern

```python
n = int(input())
for i in range(n):
    for j in range(n):
        print("*",end=' ')
    print()
```

## 5.Right Angle triangle(Number Pattern)

```python
n = int(input())
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end = " ")
    print()
```

## 6. Inverted Right Angle Triangle(Number Pattern)

```python
n = int(input())
for i in range(n,0,-1):
    for j in range(1,i+1):
        print(j,end = " ")
    print()
```
