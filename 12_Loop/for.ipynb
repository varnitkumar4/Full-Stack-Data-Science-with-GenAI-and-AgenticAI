# some function in for loop

### break

```python
for i in range(1,11):
    if i == 9:
        break
    print(i)
```

### continue

```python
for i in range(1,11):
    if i == 9:
        continue
    print(i)
```

```python
for i in range(1,11):
    if i == 5:
        continue
    print(i)
```
