```python
import pandas as pd

data = {
    "Employee": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "Sales", "IT", "Finance"],
    "Gender": ["Male", "Female", "Female", "Male", "Male", "Female", "Male", "Female"],
    "Age": [25, 30, 28, 35, 40, 32, 27, 45],
    "Salary": [40000, 45000, 50000, 60000, 55000, 48000, 52000, 70000],
    "Experience": [2, 5, 4, 10, 12, 7, 3, 15]
}

df = pd.DataFrame(data)

print(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.info()
```

# **Which department has the highest average salary?**

```python
sns.barplot(data=df,x='Department',y='Salary')
```

# **Are there any unusually high or low salaries in the dataset?**

```python
plt.figure(figsize=(8,4))
sns.boxplot(data=df,x='Salary')
plt.show()
```

# **How many employees are there in each department?**

```python
sns.countplot(data=df,x = 'Department')
plt.ylabel("Count of Employees")
plt.show()
```

# **We want to create a scatterplot of Experience vs Salary. We want each data point to be clearly visible and slightly transparent.**

```python
sns.scatterplot(data=df,x='Experience',y='Salary')
```

```python
sns.scatterplot(
    data=df,
    x="Experience",
    y="Salary",
    marker="^",
    alpha=0.6
)

plt.show()
```

# **Age vs Salary, separated by Gender, with different colors for each gender.**

```python
sns.scatterplot(
    data=df,
    x='Age',
    y = 'Salary',
    hue='Gender',
    alpha=0.9,
    marker='*',
    palette=['green','red'],
    s=200)
plt.show()
```

# **Create a scatterplot of Age vs Salary, where Gender should be represented using different marker shapes instead of different colors.**

```python
sns.scatterplot(
    data=df,
    x='Age',
    y = 'Salary',
    hue='Gender',
    alpha=0.9,
    palette=['yellow','black'],
    s=200
    ,style='Gender')
plt.show()
```

# **We want to compare the Age vs Salary relationship separately for each Gender, but instead of putting both groups on the same graph, we want separate graphs/panels.**

```python
sns.relplot(
    data=df,
    x="Age",
    y="Salary",
    col="Gender",
    kind="scatter"
)
plt.show()
```
