### 4. Logical Operators

```python
# and , or , not

l = 45
l1 = 43
print(l1<l and l>l1)
```

```python
l = 45
l1 = 43
print(l1>l and l>l1)
```

```python
l = 45
l1 = 43
print(l1>l or l>l1)
```

```python
l = 45
l1 = 43
print(not(l1>l and l>l1))
```

#### and → Both conditions must be True.                                                                          
#### or → At least one condition must be True.                                                                         
#### not → Reverses True/False.                                                                         


### 5. Membership Operators

```python
# in
# not in

name = ['rahul' , 'varnit', 'ram']
print('varnit' in name)
```

```python
name = ['rahul' , 'varnit', 'ram']
print('varnit' not in name)
```

```python
name = ['rahul' , 'varnit', 'ram']
print('singh' not in name)
```

### 6. Identity Operators

```python
# is 
# is not 

a = None

print(a is None)
```

#### ==  # checks whether values are equal
#### is  # checks whether they are the same object
