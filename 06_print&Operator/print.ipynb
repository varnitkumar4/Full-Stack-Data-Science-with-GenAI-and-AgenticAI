## print Format method 

```python
num1 = 10
num2 = 40
add = num1 + num2
print("addition of {} + {} is = {}".format(num1,num2,add))
```

```python
name = "Varnit Singh"
age = 21
city = "Muzaffarpur"
print("my name is {} and i am {} years old from {}".format(name,age,city))
```
