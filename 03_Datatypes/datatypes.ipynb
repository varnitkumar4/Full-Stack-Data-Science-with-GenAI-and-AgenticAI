# Datatypes in python 

### 1 - int (Integer)

```python
i = 5
print(i)
print(type(i))
```

```python
i1 = 34
print(i1)
```

### 2 - Float

```python
f = 5.6
print(f)
print(type(f))
```

### 3 - str (string)

```python
s = "Varnit"
print(s)
print(type(s))
```

### 4 - Bool (Boolean)

```python
b = True
print(b)
print(type(b))
```

```python
b2 = False
print(b2)
print(type(b2))
```

```python
print(id(b2)) # id give the address of the variable
```

### 5 - Complex

```python
c = 6 + 8j
print(c)
print(type(c))
```

```python
c = 6 + 8J # either small or capital J allowed
print(c)
print(type(c))
```
