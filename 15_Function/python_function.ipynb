```python
def add_sub_mul(x,y):
    c = x + y
    d = x - y
    e = x * y 
    return c,d,e 

r, r1,r2  = add_sub_mul(5,6) 

print(r)
print(r1)
print(r2)
```

# 1- positional argument

```python
def person(name, age):    
    print(name)
    print(age)    
person('nit', 23)
```

# keyword argument

```python
def person(name, age):    
    print(name)
    print(age+1)    
person(age = 23, name = 'nit')
```

```python
def person(mob,name, age):    
    print(name)
    print(age+1)    
    print(mob)
person(2345,age = 23, name = 'nit') # error-  positional argument follows keyword argument
```

```python
def person(name, age, mob, addr):    
    print(name)
    print(age+1)    
    print(mob)
person(age = 23, name = 'nit', mob=2345, Addr = 'hyd') # error - person() got an unexpected keyword argument 'Addr'. Did you mean 'addr'?
```
