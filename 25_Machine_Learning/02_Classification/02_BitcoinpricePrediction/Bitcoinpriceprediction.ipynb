```python
import yfinance as yf

btc = yf.Ticker('BTC-USD')
prices1 = btc.history(start='2021-10-09', end='2026-10-09')  # 8_OCT tak ka hi data le rhe hai , end date exclusive hoti hai.
prices1.drop(columns=['Open', 'High', 'Low', 'Dividends', 'Stock Splits'], inplace=True)

eth = yf.Ticker('ETH-USD')
prices2 = eth.history(start='2021-10-09', end='2026-10-09')
prices2.drop(columns=['Open', 'High', 'Low', 'Dividends', 'Stock Splits'], inplace=True)

usdt = yf.Ticker('USDT-USD')
prices3 = usdt.history(start='2021-10-09', end='2026-10-09')
prices3.drop(columns=['Open', 'High', 'Low', 'Dividends', 'Stock Splits'], inplace=True)

bnb = yf.Ticker('BNB-USD')
prices4 = bnb.history(start='2021-10-09', end='2026-10-09')
prices4.drop(columns=['Open', 'High', 'Low', 'Dividends', 'Stock Splits'], inplace=True)

p1 = prices1.join(prices2, lsuffix=' (BTC)', rsuffix=' (ETH)')
p2 = prices3.join(prices4, lsuffix=' (USDT)', rsuffix=' (BNB)')

data1 = p1.join(p2, lsuffix='_', rsuffix='_')

print(data1.tail())
```

```python
# Feature and target split 
# Features
X = data[['Close (BTC)', 'Volume (BTC)']]

y_day = data['Next_Day_BTC_Close']
y_7day = data['Next_7_Day_BTC_Close']

print(X.shape)
print(y_day.shape)
print(y_7day.shape)
```

```python
# Total test rows
test_size = 364
split = len(X) - test_size

# Next-day prediction
X_train_day = X.iloc[:split - 1]
X_test_day = X.iloc[split:]

y_day_train = y_day.iloc[:split - 1]
y_day_test = y_day.iloc[split:]

# Next-7-day prediction
X_train_7day = X.iloc[:split - 7]
X_test_7day = X.iloc[split:]

y_7day_train = y_7day.iloc[:split - 7]
y_7day_test = y_7day.iloc[split:]
```

```python
from sklearn.linear_model import LinearRegression

# Next-day model
lr_day = LinearRegression()
lr_day.fit(X_train_day, y_day_train)

y_day_pred = lr_day.predict(X_test_day)

# Next-7-day model
lr_7day = LinearRegression()
lr_7day.fit(X_train_7day, y_7day_train)

y_7day_pred = lr_7day.predict(X_test_7day)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Next-day model evaluation
print("Next-day Model")
print("MAE:", mean_absolute_error(y_day_test, y_day_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_day_test, y_day_pred)))
print("R2 Score:", r2_score(y_day_test, y_day_pred))

# Next-7-day model evaluation
print("\nNext-7-day Model")
print("MAE:", mean_absolute_error(y_7day_test, y_7day_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_7day_test, y_7day_pred)))
print("R2 Score:", r2_score(y_7day_test, y_7day_pred))
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
models = {
    'Linear Regression': LinearRegression(),
    'Ridge Regression': Ridge(alpha=1.0),
    'Lasso Regression': Lasso(alpha=1.0),
    'ElasticNet Regression': ElasticNet(alpha=1.0, l1_ratio=0.5),
    'SVR': SVR(kernel='rbf'),
    'Decision Tree': DecisionTreeRegressor(random_state=42),
    'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42),
    'Gradient Boosting': GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, random_state=42),
    'KNN': KNeighborsRegressor(n_neighbors=5),
    'Neural Network (MLP)': MLPRegressor(hidden_layer_sizes=(100, 50), max_iter=1000, random_state=42)
}

results = []

datasets = {
    'Next-Day': (X_train_day, X_test_day, y_day_train, y_day_test),
    'Next-7-Day': (X_train_7day, X_test_7day, y_7day_train, y_7day_test)
}

for target, (X_train, X_test, y_train, y_test) in datasets.items():
    for name, model in models.items():

        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

        results.append({
            'Target': target,
            'Model': name,
            'MAE': mean_absolute_error(y_test, y_pred),
            'RMSE': np.sqrt(mean_squared_error(y_test, y_pred)),
            'R2 Score': r2_score(y_test, y_pred)
        })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    ['Target', 'R2 Score'],
    ascending=[True, False]
)

print(results_df.round(3).to_string(index=False))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

models = {
    'SVR': make_pipeline(
        StandardScaler(),
        SVR(kernel='rbf')
    ),
    'KNN': make_pipeline(
        StandardScaler(),
        KNeighborsRegressor(n_neighbors=5)
    ),
    'MLP': make_pipeline(
        StandardScaler(),
        MLPRegressor(
            hidden_layer_sizes=(100, 50),
            max_iter=1000,
            random_state=42
        )
    )
}

datasets = {
    'Next-Day': (X_train_day, X_test_day, y_day_train, y_day_test),
    'Next-7-Day': (X_train_7day, X_test_7day, y_7day_train, y_7day_test)
}

results_scaled = []

for target, (X_train, X_test, y_train, y_test) in datasets.items():
    for name, model in models.items():
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

        results_scaled.append({
            'Target': target,
            'Model': name,
            'MAE': mean_absolute_error(y_test, y_pred),
            'RMSE': np.sqrt(mean_squared_error(y_test, y_pred)),
            'R2 Score': r2_score(y_test, y_pred)
        })

results_scaled_df = pd.DataFrame(results_scaled)

print(
    results_scaled_df
    .sort_values(['Target', 'R2 Score'], ascending=[True, False])
    .round(3)
    .to_string(index=False)
)
```

```python
# LR 
from sklearn.linear_model import LinearRegression

lr_day_all = LinearRegression()
lr_day_all.fit(X_train_day, y_day_train)
lr_day_pred_all = lr_day_all.predict(X_test_day)

lr_7day_all = LinearRegression()
lr_7day_all.fit(X_train_7day, y_7day_train)
lr_7day_pred_all = lr_7day_all.predict(X_test_7day)
```

```python
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

models = {
    'Random Forest': RandomForestRegressor(
        n_estimators=100, random_state=42
    ),
    'Gradient Boosting': GradientBoostingRegressor(
        n_estimators=100, learning_rate=0.1, random_state=42
    )
}

datasets = {
    'Next-Day': (X_train_day, X_test_day, y_day_train, y_day_test),
    'Next-7-Day': (X_train_7day, X_test_7day, y_7day_train, y_7day_test)
}

results = []

for target, (X_train, X_test, y_train, y_test) in datasets.items():
    for name, model in models.items():
        model.fit(X_train, y_train)
        pred = model.predict(X_test)

        results.append({
            'Target': target,
            'Model': name,
            'MAE': mean_absolute_error(y_test, pred),
            'RMSE': np.sqrt(mean_squared_error(y_test, pred)),
            'R2 Score': r2_score(y_test, pred)
        })

import pandas as pd

print(
    pd.DataFrame(results)
    .sort_values(['Target', 'R2 Score'], ascending=[True, False])
    .round(3)
    .to_string(index=False)
)
```
