# Python type casting

### 1 - All other datatype to Integer (int)

```python
print(int(5.7))
print(int("45"))
print(int(True))
```

```python
print(int("hi"))
```

```python
print(int(6+8j))
```

#### in type casting while typecast from all other datatype to int possible except complex & text string 

### 2 - All other datatype to float

```python
print(float(5))
print(float("5"))
print(float(True))
```

```python
print(float("Hi"))
```

```python
print(float(6+8j))
```

### 3 - All other datatype to bool

```python
print(bool(7))
print(bool(0))
print(bool("68"))
print(bool("Hi"))
print(bool(8+9j))
```

#### in bool we can convert all other datatype into bool

### 4 - All other datatype to complex

```python
print(complex(6))
print(complex(6.5))
print(complex("6"))
print(complex(True))
```

```python
print(complex("hi"))
```

```python
complex(67,"67")
```

```python
complex(90,87)
```

```python
complex("89",78)
```
