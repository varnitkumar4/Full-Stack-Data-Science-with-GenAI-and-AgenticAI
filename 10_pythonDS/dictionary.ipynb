# Dictionary 

### Create Dictionary 

```python
d = {}        # empty dictionary
print(d)
print(type(d))
```

```python
d1 = dict()   # it is also a empty dictionary
print(d1)
print(type(d1))
```

```python
d2 = {1:'one',2:'two',3:'three'}  # dictionary with integer keys 
d2
```

```python
d3 = dict({1:'one',2:'two',3:'three'}) # Create dictionary using dict()
d3
```

```python
d4 = {'a':'apple','b':'ball','c':'cat'} # dictionary with character keys
d4
```

```python
d5 = {'a':'apple','r':'rat',1:'one',2:'two'} # dictionary with mixed keys
d5
```

```python
d5
```

```python
d5.keys() # Return Dictionary Keys using keys() method 
```

```python
d5.values() # Return Dictionary Values using values() method 
```

```python
d5.items() # Access each key-value pair within a dictionary 
```

```python

d6 = {1:'one' , 2:'two' , 'A':['ram' , 'shyam' , 'krishna']} # dictionary with list
d6 
```

```python

d7 = {1:'one' , 2:'two' , 'A':['asif' , 'john' , 'Maria'],'B':('bat','ball','belt')} # dictionary with list and tuple
d7 
```

```python
keys = {'a','b','c','d'}
d7 = dict.fromkeys(keys)  # Create a dictionary from a sequence of keys
d7
```

```python
keys = {'a','b','c','d'}
value = 10
d7 = dict.fromkeys(keys,value)  # Create a dictionary from a sequence of keys
d7
```

```python
keys = {'a','b','c','d'}
value = {1,2,3,4}
d7 = dict.fromkeys(keys,value)  # Create a dictionary from a sequence of keys
d7
```

```python
keys = {'a','b','c','d'}
value = [10,29,45,3,2,34]
d7 = dict.fromkeys(keys,value)  # Create a dictionary from a sequence of keys
d7
```

```python
value.append(9)
d7
```

### Add, Remove & Change Items 

```python
dic2 = {'name':'varnit','id':22105127042,'DOB':2003,'addres':'Patna'}
dic2
```

```python
dic2['DOB'] = 2004        # Changing Dictionary Items 
dic2['addres'] = "Muzaffarpur"
dic2
```

```python
dic2
```

```python
dict1 = {'DOB':1995} 
dic2.update(dict1) 
dic2 
```

```python
dic2['job'] = 'Ai engineer'
dic2
```

```python
dic2.pop('job') # Removing items in the dictionary using Pop method
dic2
```

```python
dic2.popitem()   # A random item is removed 
```

```python
dic2
```

```python
del[dic2['id']] # Removing item using del method
```

```python
dic2
```

```python
dic2.clear()  # Delete all items of the dictionary using clear method
```

```python
dic2
```

```python
del dic2 # Delete the dictionary object
```

```python
dic2
```

### Copy Dictionary 

```python
dic3 = {'name':'varnit','id':22105127042,'DOB':2003,'addres':'Bihar'}
dic3
```

```python
dic4 = dic3  # Create a new reference "dic4"
```

```python
id(dic3) , id(dic4) # The address of both dic3 & dic4 will be the same 
```
