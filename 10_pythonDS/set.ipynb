# set operation

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}
c = {8,9,10}
```

### Union 

```python
a.union(b)
```

```python
a | b | c
```

### Intersection 

```python
print(a)
print(b)
print(c)
```

```python
a.intersection(b)
```

```python
b.intersection(c)
```

```python
a & b  # use of intersection
```

```python
a & b & c
```

```python
b.difference(c)
```
