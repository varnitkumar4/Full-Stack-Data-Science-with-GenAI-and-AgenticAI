# Comment in python 

### Single line comment

```python
# This is a comment 
name = "Varnit"  # my name 
name 
```

### Multi line comment 

```python
# This is a comment
# written on multiple lines.
# Python will ignore these lines.

name = "Varnit"
name 
```

```python
"""
This looks like a
multi-line comment.
"""

# also called Docstring
```
