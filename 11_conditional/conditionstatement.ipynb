## write a python code for even number

### if

```python
x = 4
r = x%2

if r == 0:
    print("evne number")
```

```python
x = 4
r = x%2

if r == 0:
    print("evne number")
if r != 0:
    print("odd number")
```

### if else

```python
x = 4
r = x%2

if r == 0:
    print("evne number")
else:
    print("odd number")
```

### nested if

```python
x = 4
r = x%2

if r == 0:
    print("evne number")
    if x>5:
        print("large number")
else:
    print("odd number")
```

```python
x = 4
r = x%2

if r == 0:
    print("evne number")
    if x>5:
        print("large number")
    else:
        print("small")
else:
    print("odd number")
```
